```python
#NAMA    : Aufa Bilkis Gethsa
#NIM     : 4232501031
#TANGGAL : 9/10/2026
#TUJUAN  : Melatih operator aritmatika dan perbandingan lewat perhitungan belanja dan patungan
#SOAL 1

jumlah_buku = 7
harga_buku = 18500
uang_budi = 150000
jumlah_orang = 4

jumlah_buku   = int(input("Jumlah buku     : "))
harga_buku    = int(input("Harga per buku  : "))
uang_budi     = int(input("Uang Budi       : "))
jumlah_orang  = int(input("Jumlah orang    : "))

total_belanja = jumlah_buku * harga_buku
diskon_belanja = total_belanja // 10          #10% dari total belanja
bayar = total_belanja - diskon_belanja

kembalian_budi = uang_budi - bayar

patungan_per_orang = bayar // jumlah_orang
sisa_uang_budi = bayar % jumlah_orang

cukup = uang_budi >= bayar

print()
print("Total belanja      :", total_belanja)
print("Diskon belanja     :", diskon_belanja)
print("Bayar              :", bayar)
print("Kembalian Budi     :", kembalian_budi)
print("Patungan per orang :", patungan_per_orang)
print("Sisa uang Budi     :", sisa_uang_budi)
print("Uang Budi cukup?   :", cukup)
```

```python
#NAMA    : Aufa Bilkis Gethsa
#NIM     : 4232501031
#TANGGAL : 9/10/2026
#TUJUAN  : Melatih operator pada bilangan desimal untuk menghitung energi PLTS dan kecukupannya.
#SOAL 2

jumlah_panel = 12
daya_panel = 450        # Wp per panel
jam_efektif = 5         # jam/hari
efisiensi = 0.8
kebutuhan = 25          # kWh/hari
tarif = 1444.70         # Rp/kWh
hari = 30

jumlah_panel = int(input("Jumlah panel           : "))
daya_panel = float(input("Daya per panel (Wp)    : "))
jam_efektif = float(input("Jam efektif (jam/hari) : "))
efisiensi = float(input("Efisiensi (0-1)        : "))
kebutuhan = float(input("Kebutuhan (kWh/hari)   : "))
tarif = float(input("Tarif (Rp/kWh)         : "))
hari = int(input("Jumlah hari            : "))

energi_harian = jumlah_panel * daya_panel / 1000 * jam_efektif * efisiensi
energi_bulanan = energi_harian * hari
nilai_listrik = energi_bulanan * tarif

kekurangan = kebutuhan - energi_harian
mencukupi = energi_harian >= kebutuhan

print()
print("Energi harian (kWh)   :", round(energi_harian, 1))
print("Energi bulanan (kWh)  :", round(energi_bulanan, 1))
print("Nilai listrik (Rp)    :", round(nilai_listrik, 1))
print("Kekurangan (kWh)      :", round(kekurangan, 1))
print("Mencukupi?            :", mencukupi)
```

```python
#NAMA    : Aufa Bilkis Gethsa
#NIM     : 4232501031
#TANGGAL : 9/10/2026
#TUJUAN  : Melatih operator pangkat, sisa bagi, dan logika and untuk menghitung bahan bakar generator dan rugi kabel.
#SOAL 3

daya = 200              # kW
jam = 8                 # jam/hari
konsumsi = 0.25         # liter per kWh
harga_solar = 13500     # Rp/liter
tangki = 1000           # liter
arus = 20               # A
hambatan = 0.5          # Ohm

daya = float(input("Daya generator (kW)      : "))
jam = float(input("Jam operasi (jam/hari)   : "))
konsumsi = float(input("Konsumsi (liter/kWh)     : "))
harga_solar = float(input("Harga solar (Rp/liter)   : "))
tangki = float(input("Kapasitas tangki (liter) : "))
arus = float(input("Arus kabel (A)           : "))
hambatan = float(input("Hambatan kabel (Ohm)     : "))

energi = daya * jam
solar = energi * konsumsi
biaya = solar * harga_solar

hari_cukup = tangki // solar
sisa_solar = tangki % solar

rugi_kabel = arus ** 2 * hambatan
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print()
print("Energi (kWh)     :", energi)
print("Solar (liter)    :", solar)
print("Biaya (Rp)       :", biaya)
print("Tangki cukup     :", hari_cukup, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
