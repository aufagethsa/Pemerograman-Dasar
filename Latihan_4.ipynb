```python
#NAMA: Aufa Bilkis Gethsa
#NIM: 4232501031
#TANGGAL: 3 Oktober 2026
#TUJUAN: Menyusun ekspresi logika kompleks dengan operator and, or, dan modulus (%) untuk menentukan apakah suatu tahun merupakan tahun kabisat.

tahun = int(input("Masukkan tahun : "))

kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

print("Tahun", tahun, "kabisat?", kabisat)
```

```python
#NAMA: Aufa Bilkis Gethsa
#NIM: 4232501031
#TANGGAL: 3 Oktober 2026
#TUJUAN: Menggabungkan lingkup variabel, fungsi input, fungsi output, dan operasi aritmatika dalam satu program yang utuh.

nama_toko = "Warung Es Coklat Bibibop"
pajak = 0.1   # pajak 10%

def hitung_total(harga, jumlah):
    subtotal = harga * jumlah
    total = subtotal + (subtotal * pajak)
    return total

# fungsi input
print("=== Selamat datang di", nama_toko, "===")
nama = input("Nama pelanggan  : ")
harga = float(input("Harga per porsi : "))
jumlah = int(input("Jumlah porsi    : "))
total_bayar = hitung_total(harga, jumlah)
uang = float(input("Uang dibayar    : "))
kembalian = uang - total_bayar

# fungsi output
print("\n--- Struk Pembayaran ---")
print("Pelanggan : %s" % nama)
print("Total     : Rp %.2f" % total_bayar)
print("Kembalian : Rp %.2f" % kembalian)
```
